```python
data_small = data.iloc[:5000, 0:]
data_small.head()
```

```python
data_small.info()
```

```python
smoker = data_small["Smoking_History"]
height = data_small["Height_(cm)"]

height_smoker = data_small[data_small['Smoking_History'] == 'Yes']['Height_(cm)']
height_non_smoker = data_small[data_small['Smoking_History'] == 'No']['Height_(cm)']

stat_smoker, p_value_smoker = stats.shapiro(height_smoker)
stat_non_smoker, p_value_non_smoker = stats.shapiro(height_non_smoker)


if p_value_smoker > 0.05 and p_value_non_smoker > 0.05:
  print("Podaci su normalno distribuirani. Koristit će se parametarske metode.")
else:
  print("Podaci nisu normalno distribuirani. Koristit će se neparametrijske metode.")
```

```python
stat, p = stats.mannwhitneyu(height_smoker, height_non_smoker)
p
```

ANOVA

```python
health_types = data_small['General_Health'].unique()
groups = [data_small['Height_(cm)'][data_small['General_Health'] == d] for d in health_types]

for i, health in enumerate(health_types):
    statistic, p_value = stats.shapiro(groups[i])
    if p_value > 0.05:
        print(f"Podaci za osobe sa zdravljem {health} su normalno distribuirani")
    else:
        print(f"Podaci za osobe sa zdravljem {health} nisu normalno distribuirani")

statistic, p_value = stats.levene(*groups,center='mean')

print(f"P-vrijednost za Leveneov test: {p_value}")
if p_value > 0.05:
    print("Podaci su homoscedasticni")
else:
    print("Podaci nisu homoscedasticni")
```

```python
statistic, p_value = stats.kruskal(*groups)

print(f"Kruskal-Wallis test statistic: {statistic}")
print(f"P-value: {p_value}")
```

```python
import scikit_posthocs as sp

posthoc = sp.posthoc_dunn(groups, p_adjust = 'bonferroni')
print(posthoc)
```
